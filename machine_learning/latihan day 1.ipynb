# Prediksi Nilai Ujian Berdasarkan banyaknya jam belajar

# import library & siapkan data


```python
import numpy as np
from sklearn.linear_model import LinearRegression
import matplotlib.pyplot as plt

# DATA: [Jam Belajar], [Nilai Ujian]
# Kita buat seolah-olah ada 5 siswa
jam_belajar = np.array([1, 2, 4, 6, 7, 8]).reshape(-1, 1) # Fitur (X)
nilai_ujian = np.array([50, 60, 75, 85, 95, 100])            # Target (y)

# .reshape(-1, 1) mengubah data kita jadi format yang dimengerti scikit-learn
```

# fitting model dan trainning


```python
# Buat modelnya
model = LinearRegression()

# Suruh model belajar dari data kita (training)
model.fit(jam_belajar, nilai_ujian)

print("Oke, model sudah selesai belajar!")
```

# testing

```python
# Buat prediksi berdasarkan data jam_belajar untuk menggambar garis
prediksi_nilai = model.predict(jam_belajar)

# Gambar data asli (titik-titik biru)
plt.scatter(jam_belajar, nilai_ujian, color='blue', label='Data Asli')

# Gambar garis hasil belajar model (garis merah)
plt.plot(jam_belajar, prediksi_nilai, color='red', label='Garis Prediksi Model')

# Kasih judul dan label biar grafiknya jelas
plt.title('Jam Belajar vs. Nilai Ujian')
plt.xlabel('Jam Belajar')
plt.ylabel('Nilai Ujian')
plt.legend()
plt.grid(True)
plt.show()
```

# uji dengan data baru

```python
# Siapkan data baru: 5 jam belajar
jam_belajar_baru = np.array([[5]]) # Harus dalam format 2D [[...]]

# Minta model untuk menebak nilainya
prediksi_baru = model.predict(jam_belajar_baru)

print(f"Jika belajar selama {jam_belajar_baru[0][0]} jam, prediksi nilainya adalah: {prediksi_baru[0]:.2f}")
```
